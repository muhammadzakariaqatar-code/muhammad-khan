```python
import ipywidgets as widgets
from IPython.display import display
import math

# Create the output widget
output = widgets.Output()

# Create the calculator display
display_widget = widgets.Text(
    value='',
    placeholder='0',
    description='',
    disabled=True,
    style={'description_width': 'initial'}
)

# Define the button layout
button_layout = widgets.Layout(width='50px', height='50px', margin='2px')

# Define button style
button_style = widgets.ButtonStyle(button_color='black')

# Create buttons with white text color
buttons = [
    widgets.Button(description=str(i), layout=button_layout, style=button_style, button_style='info') for i in range(10)
] + [
    widgets.Button(description='+', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='-', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='*', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='/', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='.', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='=', layout=button_layout, style=button_style, button_style='info'),
    widgets.Button(description='C', layout=button_layout, style=button_style, button_style='danger'), # Clear button
    widgets.Button(description='sqrt', layout=button_layout, style=button_style, button_style='info') # Square root button
]

# Arrange buttons in a grid-like structure
button_rows = [
    widgets.HBox([buttons[7], buttons[8], buttons[9], buttons[10]]), # 7, 8, 9, +
    widgets.HBox([buttons[4], buttons[5], buttons[6], buttons[11]]), # 4, 5, 6, -
    widgets.HBox([buttons[1], buttons[2], buttons[3], buttons[12]]), # 1, 2, 3, *
    widgets.HBox([buttons[0], buttons[14], buttons[15], buttons[13]]), # 0, ., =, /
    widgets.HBox([buttons[16], buttons[17]]) # C, sqrt
]

# Combine all elements
calculator_vbox = widgets.VBox([display_widget] + button_rows)

# Function to update the display when a button is clicked
def on_button_click(b):
    with output:
        if b.description == '=':
            try:
                display_widget.value = str(eval(display_widget.value))
            except:
                display_widget.value = 'Error'
        elif b.description == 'C':
            display_widget.value = ''
        elif b.description == 'sqrt':
            try:
                value = float(display_widget.value)
                if value >= 0:
                    display_widget.value = str(math.sqrt(value))
                else:
                    display_widget.value = 'Error: Negative input for sqrt'
            except:
                display_widget.value = 'Error'
        else:
            display_widget.value += b.description

# Attach the click event to each button
for btn in buttons:
    btn.on_click(on_button_click)

# Display the calculator and output area
display(calculator_vbox, output)
```
